```python
# Remove unwanted Unnamed columns
df = df.loc[:, ~df.columns.str.startswith("Unnamed")]

print("Shape after removing unwanted columns:", df.shape)

print("\nRemaining columns:")
print(df.columns.tolist())
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], format="%d-%b-%y", errors="coerce")

# Sort data by date
df = df.sort_values("Date").reset_index(drop=True)
```

```python
print("\nFirst 5 dates:")
print(df["Date"].head())

print("\nLast 5 dates:")
print(df["Date"].tail())

print("\nDate range:")
print(df["Date"].min(), "to", df["Date"].max())
```

```python
# Columns that should contain numerical values
numeric_columns = ["Prev Close","Open Price","High Price","Low Price","Last Price","Close Price","Average Price","Total Traded Quantity",
                   "Turnover ₹","No. of Trades","Deliverable Qty","% Dly Qt to Traded Qty"]

# Remove commas and convert to numeric
for col in numeric_columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Check data types
print("Data types after conversion:\n")
print(df.dtypes)
```

```python
print("\nMissing values after numeric conversion:")
print(df.isnull().sum())
```

```python
# Check duplicate dates
duplicate_dates = df[df["Date"].duplicated(keep=False)].sort_values("Date")

print("Number of duplicate date rows:", len(duplicate_dates))
print("Number of unique duplicate dates:", duplicate_dates["Date"].nunique())

print("\nDuplicate dates:")
print(duplicate_dates["Date"].value_counts().sort_index())
```

```python
# Identify duplicate dates
duplicate_mask = df["Date"].duplicated(keep=False)

# Remove suspicious duplicate records having only 1 trade
df_clean = df[~(duplicate_mask & (df["No. of Trades"] == 1))].copy()

# Reset index
df_clean.reset_index(drop=True, inplace=True)

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nRows removed:", len(df) - len(df_clean))

print("\nRemaining duplicate dates:",df_clean["Date"].duplicated().sum())
```

```python
# Remove the remaining suspicious duplicate record
df_clean = df_clean[
    ~(
        (df_clean["Date"] == pd.Timestamp("2019-02-25")) &
        (df_clean["No. of Trades"] == 2)
    )
].copy()

# Reset index
df_clean.reset_index(drop=True, inplace=True)

print("Cleaned shape:", df_clean.shape)
print("Remaining duplicate dates:", df_clean["Date"].duplicated().sum())
```

## Relative Strength Index

```python
delta = df_clean["Close Price"].diff()

gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

avg_gain = gain.rolling(14).mean()
avg_loss = loss.rolling(14).mean()

rs = avg_gain / avg_loss

df_clean["RSI_14"] = 100 - (100 / (1 + rs))

print(df_clean[["Date", "Close Price", "RSI_14"]].tail())
```

```python
import requests
import pandas as pd
import re

results = []

for i, row in llm_input.iterrows():

    prompt = f"""
You are analyzing TCS (Tata Consultancy Services) stock.

Based only on the news below, predict the likely movement
of TCS stock on the next trading day.

Return ONLY these 4 lines:

Movement: UP/DOWN/SAME
P(UP): 0.xx
P(DOWN): 0.xx
P(SAME): 0.xx

The three probabilities must add up to 1.

News:
{row["headline"]}
"""

    data = {
        "model": "llama3.2:3b",
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(
        "http://localhost:11434/api/generate",
        json=data
    )

    output = response.json()["response"]

    movement = re.search(r"Movement:\s*(UP|DOWN|SAME)", output)
    p_up = re.search(r"P\(UP\):\s*([0-9.]+)", output)
    p_down = re.search(r"P\(DOWN\):\s*([0-9.]+)", output)
    p_same = re.search(r"P\(SAME\):\s*([0-9.]+)", output)

    results.append({
        "Date": row["Date"],
        "Movement": movement.group(1) if movement else None,
        "P_UP": float(p_up.group(1)) if p_up else None,
        "P_DOWN": float(p_down.group(1)) if p_down else None,
        "P_SAME": float(p_same.group(1)) if p_same else None
    })

llm_features = pd.DataFrame(results)

print(llm_features)
```

```python
from sklearn.preprocessing import StandardScaler

scaler_return = StandardScaler()

X_train_return_scaled = scaler_return.fit_transform(X_train_return)
X_test_return_scaled = scaler_return.transform(X_test_return)

print("X_train_scaled:", X_train_return_scaled.shape)
print("X_test_scaled:", X_test_return_scaled.shape)
```

```python
window = 20

X_train_seq_return = []
y_train_seq_return = []

for i in range(window, len(X_train_return_scaled)):
    X_train_seq_return.append(
        X_train_return_scaled[i-window:i]
    )
    y_train_seq_return.append(
        y_train_return.iloc[i]
    )

X_train_seq_return = np.array(X_train_seq_return)
y_train_seq_return = np.array(y_train_seq_return)

print("X_train_seq:", X_train_seq_return.shape)
print("y_train_seq:", y_train_seq_return.shape)
```

```python
X_test_seq_return = []
y_test_seq_return = []

for i in range(window, len(X_test_return_scaled)):
    X_test_seq_return.append(
        X_test_return_scaled[i-window:i]
    )
    y_test_seq_return.append(
        y_test_return.iloc[i]
    )

X_test_seq_return = np.array(X_test_seq_return)
y_test_seq_return = np.array(y_test_seq_return)

print("X_test_seq:", X_test_seq_return.shape)
print("y_test_seq:", y_test_seq_return.shape)
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.layers import MultiHeadAttention
from tensorflow.keras.layers import LayerNormalization
from tensorflow.keras.layers import GlobalAveragePooling1D

inputs = Input(shape=(20, 17))

attention = MultiHeadAttention(
    num_heads=2,
    key_dim=16
)(inputs, inputs)

x = LayerNormalization()(attention)

x = GlobalAveragePooling1D()(x)

x = Dense(32, activation="relu")(x)

x = Dropout(0.2)(x)

outputs = Dense(1)(x)

return_model = Model(
    inputs=inputs,
    outputs=outputs
)

return_model.compile(
    optimizer="adam",
    loss="mse"
)

return_model.summary()
```

```python
history_return = return_model.fit(
    X_train_seq_return,
    y_train_seq_return,
    epochs=30,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)
```

```python
y_pred_return = return_model.predict(X_test_seq_return)

y_pred_return = y_pred_return.flatten()

print("Prediction shape:", y_pred_return.shape)
print("First 10 predicted returns:")
print(y_pred_return[:10])
```

```python
# Remove the date that Llama did not use
mistral_common = mistral_features[
    mistral_features["Date"].isin(llm_features["Date"])
].copy()

# Remove rows where movement was not generated properly
mistral_common = mistral_common.dropna(
    subset=["Movement", "P_UP", "P_DOWN", "P_SAME"]
).reset_index(drop=True)

print("Mistral common rows:", len(mistral_common))
print(mistral_common)
```

```python
# Keep only dates present in both models
ensemble = llm_features.merge(
    mistral_features,
    on="Date",
    suffixes=("_Llama", "_Mistral")
)

# Average the probabilities
ensemble["P_UP"] = (
    ensemble["P_UP_Llama"] +
    ensemble["P_UP_Mistral"]
) / 2

ensemble["P_DOWN"] = (
    ensemble["P_DOWN_Llama"] +
    ensemble["P_DOWN_Mistral"]
) / 2

ensemble["P_SAME"] = (
    ensemble["P_SAME_Llama"] +
    ensemble["P_SAME_Mistral"]
) / 2

# Final ensemble movement
ensemble["Movement"] = ensemble[
    ["P_UP", "P_DOWN", "P_SAME"]
].idxmax(axis=1).str.replace("P_", "")

print(
    ensemble[
        ["Date", "P_UP", "P_DOWN", "P_SAME", "Movement"]
    ]
)
```
